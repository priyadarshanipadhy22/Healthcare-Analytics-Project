```python
#create new colum df_clean
df_clean = df.copy()
```

```python
df_clean = df_clean.drop_duplicates()
```

```python
df_clean.columns.to_list()#to_list() show give output by list by array#df.()gives op by array by list#
```

```python
for col in df_clean.select_dtypes(include="object"):
                    #object which is no numerical categorical
    df_clean[col] = df_clean[col].str.strip()#used for remove extra space of column name
```

```python
#df_clean["Gender"].value_counts() 
df.value_counts("Gender")
```

```python
df_clean["Gender"] = (
    df_clean["Gender"]
    .str.lower()
    .str.capitalize()
)
```

```python
df_clean["Admission_Date"] = pd.to_datetime(   #change data type of date used for while #df.int64
    df_clean["Admission_Date"]
)


df_clean["Discharge_Date"] = pd.to_datetime(
    df_clean["Discharge_Date"]
)
```
