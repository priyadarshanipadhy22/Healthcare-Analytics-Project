```python
import pandas as pd
import numpy as np
import random
from faker import Faker
from datetime import datetime, timedelta

fake = Faker("en_IN")
```

```python
# Cities where patients come from

cities = [
    "Bhubaneswar",
    "Cuttack",
    "Puri",
    "Sambalpur",
    "Rourkela",
    "Berhampur",
    "Balasore",
    "Delhi",
    "Mumbai",
    "Bangalore",
    "Hyderabad",
    "Chennai",
    "Kolkata"
]


# List of diseases

diseases = [
    "Diabetes",
    "Hypertension",
    "Heart Disease",
    "Asthma",
    "Cancer",
    "COVID-19",
    "Pneumonia",
    "Kidney Stone",
    "Migraine",
    "Fracture",
    "Dengue",
    "Typhoid"
]


# Hospital departments

departments = [
    "Cardiology",
    "Neurology",
    "Orthopedics",
    "General Medicine",
    "Oncology",
    "Pediatrics",
    "Gynecology",
    "Pulmonology",
    "Nephrology"
]


# Doctors list

doctors = [
    "Dr. Amit Kumar",
    "Dr. Priya Sharma",
    "Dr. Rajesh Das",
    "Dr. Neha Singh",
    "Dr. Vivek Mishra",
    "Dr. Anjali Sahu",
    "Dr. Arjun Rao",
    "Dr. Sneha Patel",
    "Dr. Mohan Roy",
    "Dr. Pooja Verma"
]


# Blood groups

blood_groups = [
    "A+",
    "A-",
    "B+",
    "B-",
    "AB+",
    "AB-",
    "O+",
    "O-"
]


# Payment methods

payment_modes = [
    "Cash",
    "UPI",
    "Credit Card",
    "Debit Card",
    "Insurance"
]


# Insurance companies

insurance_companies = [
    "Star Health",
    "Niva Bupa",
    "ICICI Lombard",
    "HDFC ERGO",
    "Care Health",
    "None"
]


print("All master lists created successfully!")
```

```python
print(cities)
```

```python
columns = [
    "Patient_ID",
    "Patient_Name",
    "Gender",
    "Age",
    "Blood_Group",
    "City",
    "Disease",
    "Department",
    "Doctor",
    "Admission_Date",
    "Discharge_Date",
    "Hospital_Stay",
    "Surgery",
    "Surgery_Type",
    "Medicine_Cost",
    "Lab_Charge",
    "Room_Charge",
    "Doctor_Fee",
    "Total_Bill",
    "Amount_Paid",
    "Pending_Amount",
    "Payment_Mode",
    "Feedback_Rating",
    "Insurance",
    "Insurance_Provider"
]
```

```python
records = []


num_records = 10000


for i in range(num_records):


    patient_id = f"P{i+1:05d}"


    name = fake.name()


    gender = random.choice(
        ["Male","Female"]
    )


    age = random.randint(
        1,90
    )


    blood = random.choice(
        blood_groups
    )


    city = random.choice(
        cities
    )


    # Disease selection

    disease = random.choice(
        list(disease_department.keys())
    )


    # Automatic department

    department = disease_department[disease]


    # Automatic doctor

    doctor = random.choice(
        doctor_mapping[department]
    )


    # Dates

    admission = (
        datetime(2025,1,1)
        +
        timedelta(
            days=random.randint(0,364)
        )
    )


    stay=random.randint(1,10)


    discharge = (
        admission
        +
        timedelta(days=stay)
    )


    # Surgery

    surgery_type = surgery_mapping[disease]


    if surgery_type=="None":
        surgery="No"
    else:
        surgery="Yes"



    # Bill calculation

    total_bill=random.randint(
        cost_mapping[disease][0],
        cost_mapping[disease][1]
    )


    medicine=int(total_bill*0.30)

    room=int(total_bill*0.35)

    doctor_fee=int(total_bill*0.20)


    lab=(
        total_bill
        -
        medicine
        -
        room
        -
        doctor_fee
    )


    amount_paid=random.randint(
        int(total_bill*0.60),
        total_bill
    )


    pending_amount=(
        total_bill
        -
        amount_paid
    )


    feedback=random.randint(
        1,5
    )


    insurance=random.choice(
        ["Yes","No"]
    )


    if insurance=="Yes":

        provider=random.choice(
            insurance_companies[:-1]
        )

    else:

        provider="None"



    records.append([

        patient_id,
        name,
        gender,
        age,
        blood,
        city,
        disease,
        department,
        doctor,
        admission.date(),
        discharge.date(),
        stay,
        surgery,
        surgery_type,
        medicine,
        lab,
        room,
        doctor_fee,
        total_bill,
        amount_paid,
        pending_amount,
        random.choice(payment_modes),
        feedback,
        insurance,
        provider

    ])


print("10000 records generated")
```

```python
df = pd.DataFrame(records, columns=columns)
```
