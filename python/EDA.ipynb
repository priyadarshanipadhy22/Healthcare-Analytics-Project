```python
#load clean data
df = pd.read_csv(
    "../dataset/Cleaned_Hospital_Management_Records.csv"
)

df.head()
```

```python
#Convert Date Column
df["Admission_Date"] = pd.to_datetime(
    df["Admission_Date"]
)
```

```python
df.info()
```

```python
#Which month has highest patient admissions?"

df["Month"] = (
    df["Admission_Date"]
    .dt.month_name()
)
```

```python
#check
df["Month"].head()
```

```python
#Count Patients Month Wise
monthly_patients = (
    df["Month"]
    .value_counts()
)

monthly_patients
```

```python
#visualization
plt.figure(figsize=(10,5))


sns.barplot(
    x=monthly_patients.index,
    y=monthly_patients.values
)


plt.xticks(rotation=45)


plt.title(
    "Monthly Patient Admissions"
)


plt.xlabel(
    "Month"
)


plt.ylabel(
    "Number of Patients"
)


plt.show()
```

```python
#"Which month generated highest revenue?"
monthly_revenue = (
    df.groupby("Month")
    ["Total_Bill"]
    .sum()
    .sort_values(
        ascending=False
    )
)

monthly_revenue
```

```python
#chart
plt.figure(figsize=(10,5))


sns.barplot(
    x=monthly_revenue.index,
    y=monthly_revenue.values
)


plt.xticks(rotation=45)


plt.title(
    "Monthly Revenue Trend"
)


plt.show()
```

```python
###Correlation Analysis

#Find relationship between:

#Age
#Bill
#Stay
#Feedback

numeric_df = df.select_dtypes(
    include="number"
)


corr = numeric_df.corr()


corr
```
